```python
import pandas as pd 
import numpy as np
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.metrics import accuracy_score,f1_score,confusion_matrix,classification_report 
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import StandardScaler
import joblib
from math import sqrt
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
```

```python
df=pd.read_csv('titanic.csv',encoding='unicode_escape')

df2=pd.get_dummies(df[['Name','Sex','Ticket','Cabin','Embarked']]).astype(int)
df3=pd.concat([df,df2],axis=1)
df4=df3.drop(columns=['Name','Sex','Ticket','Cabin','Embarked'])

print(df4)
#determine if there is NaN
df4.isnull().sum()

#clear the nan
df4['Age']=df4['Age'].fillna(int(df4['Age'].mean(skipna=True)))

#the data is ready now
#split the data
X=df4.drop(columns='Survived')
y=df4['Survived']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)

#train model with decision tree
model1=DecisionTreeClassifier()
model1.fit(X_train,y_train)
predict1=model1.predict(X_test)

#test decision tree 
print('decision tree accuracy : ',accuracy_score(y_test,predict1))


#scaling
sc_X=StandardScaler()
X_train_sc=sc_X.fit_transform(X_train)
X_test_sc=sc_X.transform(X_test)

#train model with KNN
k=int(sqrt(len(y_test)))
model2=KNeighborsClassifier(n_neighbors=k)
model2.fit(X_train_sc,y_train)
predict2=model2.predict(X_test_sc)

#test KNN
print('KNN accuracy : ',accuracy_score(y_test,predict2))

#train model with naive bayes
model3=GaussianNB()
model3.fit(X_train,y_train)
predict3=model3.predict(X_test)

#test naive bayes
print('naive bayes accuracy : ',accuracy_score(y_test,predict3))

#train model with random forest
model4=RandomForestClassifier()
model4.fit(X_train,y_train)
predict4=model4.predict(X_test)

#test random forest
print('random forest accuracy : ',accuracy_score(y_test,predict4))

#train logistic regression 
model5=LogisticRegression()
model5.fit(X_train,y_train)
predict5=model5.predict(X_test)

#test logistic regression
print('logistic regression accuracy : ',accuracy_score(y_test,predict5))

#train model with svm 
```
